```python
def get_neighbors(state):
    zero = state.index(0)
    row, col = divmod(zero, 3)
    neighbors = []

    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:
        nr, nc = row + dr, col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_pos = nr * 3 + nc
            board = list(state)
            board[zero], board[new_pos] = board[new_pos], board[zero]
            neighbors.append(tuple(board))

    return neighbors


def dfs(initial, goal, max_depth=50):
    visited = set()
    path = [initial]

    def explore(state, depth):
        if state == goal:
            return True
        if depth >= max_depth:
            return False

        visited.add(state)

        for neighbor in get_neighbors(state):
            if neighbor not in visited:
                path.append(neighbor)

                if explore(neighbor, depth + 1):
                    return True

                path.pop()

        return False

    if explore(initial, 0):
        return path

    return None


def print_solution(path):
    if path is None:
        print("No solution found (or limit reached).")
        return

    print("Solution found in", len(path) - 1, "moves.")

    for step, state in enumerate(path):
        print(f"\nStep {step}:")
        for i in range(0, 9, 3):
            print(*state[i:i + 3])


#initial = (1, 2, 3, 0, 4, 6, 7, 5, 8)
#goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)
initial = input("Enter the initial state: ").split()
initial = tuple(map(int, initial))
goal = input("Enter the goal state: ").split()
goal = tuple(map(int, goal))

solution = dfs(initial, goal)
print_solution(solution)
```

```python
def depth_limited_search(state, goal, limit, path, path_set):
    if state == goal:
        return path[:]

    if limit == 0:
        return None

    for neighbor in get_neighbors(state):
        if neighbor not in path_set:
            path.append(neighbor)
            path_set.add(neighbor)

            result = depth_limited_search(
                neighbor, goal, limit - 1, path, path_set
            )

            if result is not None:
                return result

            path.pop()
            path_set.remove(neighbor)

    return None


def ids(initial, goal, max_limit=30):
    limit = 0

    while limit <= max_limit:
        path = [initial]
        path_set = {initial}

        result = depth_limited_search(
            initial, goal, limit, path, path_set
        )

        if result is not None:
            return result

        limit += 1

    return None


def print_solution(path):
    if path is None:
        print("No solution found within the limit.")
        return

    print("Solution found in", len(path) - 1, "moves.")

    for step, state in enumerate(path):
        print(f"\nStep {step}:")
        for i in range(0, 9, 3):
            print(*state[i:i + 3])


#initial = (1, 2, 3, 0, 4, 6, 7, 5, 8)
#goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)
initial = input("Enter the initial state: ").split()
initial = tuple(map(int, initial))
goal = input("Enter the goal state: ").split()
goal = tuple(map(int, goal))

solution = ids(initial, goal)
print_solution(solution)
```
