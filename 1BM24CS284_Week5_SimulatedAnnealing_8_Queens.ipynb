### Solving the 8 Queens Problem using Simulated Annealing

In the 8 Queens problem, we want to place 8 queens on an 8x8 chessboard such that no two queens threaten each other. This means no two queens can share the same row, column, or diagonal.

We will use **Simulated Annealing**:
- **State Representation**: A list of length 8 where the index represents the column and the value represents the row of the queen in that column.
- **Energy (E)**: The number of attacking pairs of queens (we want to minimize this, so goal is $E = 0$).
- **Transition**: Move a random queen to a different row in its column.
- **Acceptance Probability**: If the new state is better ($\\Delta E < 0$), we accept it. Otherwise, we accept it with probability $P = e^{-\\Delta E / T}$ where $\\Delta E = E_{new} - E_{current} > 0$.

```python
import random
import math

def calculate_energy(state):
    """Calculates the number of attacking pairs of queens."""
    collisions = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Check if they are in the same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                collisions += 1
    return collisions

def get_neighbor(state):
    """Generates a neighbor state by moving one random queen to a random row."""
    neighbor = list(state)
    col = random.randint(0, 7)
    row = random.randint(0, 7)
    while row == state[col]:
        row = random.randint(0, 7)
    neighbor[col] = row
    return neighbor

def simulated_annealing(initial_temp=100.0, cooling_rate=0.995, max_iterations=10000):
    # Initialize state (one queen per column, random rows)
    current_state = [random.randint(0, 7) for _ in range(8)]
    current_energy = calculate_energy(current_state)

    temp = initial_temp

    for iteration in range(max_iterations):
        if current_energy == 0:
            return current_state, iteration

        neighbor = get_neighbor(current_state)
        neighbor_energy = calculate_energy(neighbor)

        delta_e = neighbor_energy - current_energy

        # If the neighbor is better, or accepted via probability formula P = e^(-deltaE / T)
        if delta_e < 0:
            current_state = neighbor
            current_energy = neighbor_energy
        else:
            # delta_e is positive, we want P = exp(-delta_e / T)
            probability = math.exp(-delta_e / temp)
            if random.random() < probability:
                current_state = neighbor
                current_energy = neighbor_energy

        # Cool down
        temp *= cooling_rate
        if temp < 1e-6:
            break

    return current_state, iteration

def print_board(state):
    for row in range(8):
        line = ""
        for col in range(8):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

# Run the solver
solution, iterations = simulated_annealing()
print(f"Solution found in {iterations} iterations:")
print(f"State representation: {solution}")
print(f"Number of conflicts: {calculate_energy(solution)}\n")
print_board(solution)
```
