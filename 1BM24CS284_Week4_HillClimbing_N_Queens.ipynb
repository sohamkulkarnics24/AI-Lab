```python
import random

def get_conflicts(state):
    """Calculate the number of mutually attacking pairs of queens."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Check if they are in the same row or same diagonals
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_neighbors(state):
    """Generate all possible neighbors by swapping the row positions of 2 queens."""
    neighbors = []
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Create a neighbor by swapping the positions of queen i and queen j
            neighbor = list(state)
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbors.append(neighbor)
    return neighbors

def hill_climbing(n, initial_state=None):
    """Standard Hill Climbing search for N-Queens starting from a given state."""
    if initial_state is not None:
        current_state = list(initial_state)
    else:
        current_state = list(range(n))
        random.shuffle(current_state)

    current_conflicts = get_conflicts(current_state)

    while True:
        neighbors = get_neighbors(current_state)
        if not neighbors:
            break

        # Find the neighbor with the minimum conflicts
        neighbor_conflicts = [(neighbor, get_conflicts(neighbor)) for neighbor in neighbors]
        neighbor_conflicts.sort(key=lambda x: x[1])

        best_neighbor, best_conflicts = neighbor_conflicts[0]

        # If the best neighbor is not better than the current state, we have reached a local minimum
        if best_conflicts >= current_conflicts:
            break

        current_state = best_neighbor
        current_conflicts = best_conflicts

    return current_state, current_conflicts

def hill_climbing_with_random_restart(n, max_restarts=100):
    """Hill Climbing with Random Restarts to find a global optimum (0 conflicts)."""
    for i in range(max_restarts):
        state, conflicts = hill_climbing(n)
        if conflicts == 0:
            print(f"Solution found after {i+1} restarts!")
            return state
    print("Failed to find a solution within the maximum restarts.")
    return state
```

```python
def print_board(state):
    """Helper function to visualize the board state."""
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

# Get N and initial state from user
try:
    user_input = input("Enter the number of queens (N): ")
    N = int(user_input)
    if N < 1:
        raise ValueError("N must be a positive integer.")

    state_input = input(f"Enter initial state as {N} comma-separated row integers (e.g., 0,1,2,3) or leave blank for random: ").strip()

    if state_input:
        initial_state = [int(x.strip()) for x in state_input.split(",")]
        if len(initial_state) != N:
            raise ValueError(f"Initial state must have exactly {N} values.")
        for val in initial_state:
            if val < 0 or val >= N:
                raise ValueError(f"Row index must be between 0 and {N-1}.")

        print(f"\nSolving {N}-Queens problem starting from custom state: {initial_state}...")
        solution, conflicts = hill_climbing(N, initial_state=initial_state)
        print(f"Final state (conflicts={conflicts}): {solution}")
        print("\nVisualized Board:")
        print_board(solution)
    else:
        print(f"\nSolving {N}-Queens problem starting from a random state with restarts...")
        solution = hill_climbing_with_random_restart(N)
        print(f"Final state: {solution}")
        print("\nVisualized Board:")
        print_board(solution)
except ValueError as e:
    print(f"Invalid input: {e}")
```
